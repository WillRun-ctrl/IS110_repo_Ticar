#### Challenge 1

Write a program which repeatedly reads numbers until the user enters "done". Once "done" is entered, print out the total, count, and average of the numbers. If the user enters anything other than a number, detect their mistake using try and except and print an error message and skip to the next number.

    Enter a number: 4
    Enter a number: 5
    Enter a number: bad data
    Invalid input
    Enter a number: 7
    Enter a number: done
    16 3 5.333333333333333
    
[PEW] E5.1 / Answer Below:

````python
total = 0
count = 0

while True:
    users_input = input("Enter number & type 'd' if you're done: ")

    if users_input.lower() == 'd':
        break

    try:
      number = float(users_input)
      total = total + number
      count = count + 1
    except:
      print("Input Error")
      continue

if count > 0:
    average = total / count
else:
    average = 0

print(total, count, average)
````

#### Challenge 2  

Write another program that prompts for a list of numbers as above and at the end prints out both the maximum and minimum of the numbers instead of the average.

[PEW] E5.2 / Answer Below:

````python
max = None
min = None

while True:
    users_input = input("Enter number & type 'd' if you're done: ")

    if users_input.lower() == 'd':
        break

    try:
      number = float(users_input)
      total = total + number
      count = count + 1
    except:
      print("Input Error")
      continue

    if max is None:
        max = number
    elif number > max:
        max = number

    if min is None:
        min = number
    elif number < min:
        min = number

print("Max: ", max)
print("Min: ", min)
````
