**Exercise A**

```python
import numpy as np

# I made a nested list of piles of numbers in rows.
data = [ [1, 5, 10, 15], [20, 25, 30, 35], [40, 45, 50, 55] ]

# I turned the regular list into a NumPy array. This will oragnize the messy pile
B = np.array(data)

# This will show the row and column. Ex. 3-Tall 4-Wide
print(B.shape)

# Using the addition method. This will Add all the data.
print(B.sum())

# This will scan the whole file/data and find the biggest one
print(np.max(B))
```

***Exercise B***

```python
import numpy as np

B = np.ones((5, 5), dtype=int)   # It will create a 5x5 array filled with 1s. Ex. A 5x5 Box full of one 1s

print(B.shape)   # THis will show the shape of how many rows and columns has (5 rows, 5 columns)

print(B.size)    # It shows the total number of elements. 5 x 5 = 25

print(B.ndim)    # Will show the number of dimensions = 2D (rows and columns)
```

```python
import numpy as np

# This will create an array of numbers from -5 to +5
C = np.arange(-5, 6)    # np.arange will arrange where to (start, stop) and creates numbers from -5 to 5 (6 is not included because it does not have "=")

# It'll multiply every number in the array by 2
C = C * 2

# This show the resulting array
print(C)
```
